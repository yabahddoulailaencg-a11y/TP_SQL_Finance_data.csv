```python
import pandas as pd

url = 'https://raw.githubusercontent.com/yabahddoulailaencg-a11y/TP_SQL_Finance_data.csv/refs/heads/main/Finance_data.csv'

df = pd.read_csv(url)
display(df.head(10))
print('Dimensions :', df.shape)
print('Colonnes :', df.columns.tolist())

import sqlite3

conn = sqlite3.connect('tp_sql.db')
df.to_sql('donnees', conn, if_exists='replace', index=False)

display(pd.read_sql_query('SELECT * FROM donnees LIMIT 10;', conn))
display(pd.read_sql_query('SELECT COUNT(*) AS total FROM donnees;', conn))

conn.close()
```
